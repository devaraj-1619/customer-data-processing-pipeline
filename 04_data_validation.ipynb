```python
# Task 1: Validate Bronze record count
bronze_df=spark.table("customer_pipeline.customer_data.bronze_customers")
print("Bronze_count:",bronze_df.count())
```

```python
# Task 2: Validate Silver record count
silver_df=spark.table("customer_pipeline.customer_data.silver_customers")
print("silver_count:",silver_df.count())
```

```python
# Task 3: Check duplicate customer IDs
from pyspark.sql.functions import count

silver_df.groupBy("customer_id") \
    .agg(count("*").alias("total")) \
    .filter("total > 1") \
    .show()
```

```python
# Task 4: Check NULL city values
from pyspark.sql.functions import col

silver_df.filter(col("city").isNull()).show()
```

```python
# Task 5: Check invalid emails
silver_df.filter(
    col("email_status") == "invalid"
).show()
```

```python
# Task 6: Validate rejected records
rejected_df=spark.table("customer_pipeline.customer_data.rejected_customers")
print("rejected_count:",rejected_df.count())
```

```python
# Task 7: Validate total records
print("Bronze:", bronze_df.count())
print("Silver:", silver_df.count())
print("Rejected:", rejected_df.count())

print(
    "Silver + Rejected:",
    silver_df.count() + rejected_df.count()
)
```
