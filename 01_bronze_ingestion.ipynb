```python
# Reading file

df=spark.read.format("csv")\
.option("header","true")\
.option("inferschema","true")\
.load("/Volumes/customer_pipeline/customer_data/customer_volume")
```

```python
display(df)
```

```python
df.printSchema()
```

```python
print("Total_records : ",df.count())
```

```python
# save it into table called bronze_customers


df.write.format("delta")\
    .mode("overwrite")\
    .saveAsTable("customer_pipeline.customer_data.bronze_customers")
```
